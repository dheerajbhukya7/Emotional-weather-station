```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('train.txt', sep=';', header=None, names=['text','emotion'])
```

```python
uniqe_emotions = df['emotion'].unique()
emotion_number = {}
i = 0
for emotions in uniqe_emotions:
  emotion_number[emotions] = i
  i +=1


df['emotion'] = df['emotion'].map(emotion_number)
```

Lower casing

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

Remove puncuation



string methods..

```python
import string

def remove_puncuation(text):
  return text.translate(str.maketrans('', '', string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_puncuation)
```

Remove numbers

```python
def remove_numbers(text):
  new = ""
  for i in text:
    if not i.isdigit():
      new += i

      return new

  df['text'] = df['text'].apply(remove_numbers)
```

Remove URLS/links AND HTML TAGS

```python
import re

def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

```

REmove emoji & special charecters


ASKAI VALUES{preffred }


UNique code

```python
def remove_emojis(text):
    new = ""

    for i in text:
        if i.isascii():
            new += i

    return new

df['text'] =df['text'].apply(remove_emojis)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
from nltk import word_tokenize
nltk.download('punkt_tab')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
df.loc[1]['text']
```

i will conver all stopwords in to a array




Tokenization: Each individual word or symbol in the text is called a token.
EX;I LOVE NLP > ["I","LOVE","NLP"]

```python
def  remove(text):
  words = word_tokenize(text)
  new = []
  for word in words:
    if word not in stop_words:
      new.append(word)

  return " ".join(new)
```

```python
nltk.download('punkt_tab')
df['text'] = df['text'].apply(remove)
```

Train test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

VECTORIZATION


```python
from sklearn.feature_extraction.text import CountVectorizer

```

```python
bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)
y_pred = nb_model.predict(X_test_bow)
accuracy = accuracy_score(y_test, y_pred)
```

```python
nb_model.score(X_test_bow,y_test)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model_2 = MultinomialNB()
nb_model_2.fit(X_train_tfidf, y_train)
y_pred = nb_model_2.predict(X_test_tfidf)
accuracy = accuracy_score(y_test, y_pred)
```

```python
nb_model_2.score(X_test_tfidf,y_test)
```

Logistic regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
logistic_model = LogisticRegression(max_iter=1000)
logistic_model.fit(X_train_tfidf, y_train)
y_pred = logistic_model.predict(X_test_tfidf)
accuracy = accuracy
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
